```python
import sys
sys.path.append("..")

import joblib
import pandas as pd
from src.anomaly import detect, ANOMALY_FEATURES
from src.features import engineer_features

# Re-define the variables this cell needs
df = pd.read_csv("../ai4i+2020+predictive+maintenance+dataset/ai4i2020.csv", encoding="utf-8-sig")

healthy  = {"air_temp": 300.0, "process_temp": 310.5, "rpm": 1500, "torque": 40.0, "tool_wear": 60,  "type": "L"}
stressed = {"air_temp": 301.5, "process_temp": 313.5, "rpm": 1340, "torque": 68.0, "tool_wear": 215, "type": "L"}

print("Healthy :", detect(healthy))
print("Stressed:", detect(stressed))

art = joblib.load("../models/anomaly_model.joblib")
X_all = engineer_features(df)[ANOMALY_FEATURES]
flag = art["pipeline"].decision_function(X_all) < art["threshold"]
print("Flagged among normal rows :", round(flag[(df["Machine failure"] == 0).values].mean() * 100, 1), "%")
print("Flagged among failure rows:", round(flag[(df["Machine failure"] == 1).values].mean() * 100, 1), "%")
```
