```python
from src.simulator import EquipmentSimulator
from src.predict import MODEL, readings_to_features, score_to_risk, THRESHOLD
from src.anomaly import detect

print("Model threshold:", round(THRESHOLD, 3))

sim = EquipmentSimulator("TEST", "L", 60, seed=2)
sim.start()
rows = []
for i in range(70):
    r = sim.step()
    score = MODEL.predict_proba(readings_to_features(pd.DataFrame([r])))[:, 1][0]
    a = detect(r)
    rows.append({"step": i + 1, "degr": round(sim.degradation, 2), "raw_score": round(score, 3),
                 "risk": round(float(score_to_risk(score)), 1),
                 "anomaly": a["is_anomaly"], "anom_score": a["score"]})
pd.DataFrame(rows).iloc[34:62]
```
